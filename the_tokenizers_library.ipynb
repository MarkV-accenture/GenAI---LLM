# The 🤗 Tokenizers Library
We'll train a Byte-Pair Encoding (BPE) tokenizer from scratch on a tiny custom corpus and then use it to encode text.
We'll also look at special tokens and decoding.

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# tiny corpus (you can replace with your domain text)
corpus = [
    "Transformers are changing natural language processing.",
    "Generative AI enables new applications like chatbots and copilots.",
    "Tokenization breaks text into subword units.",
]
with open('corpus.txt','w') as f:
    for line in corpus:
        f.write(line+'\n')

# init tokenizer + trainer
tokenizer = Tokenizer(BPE())
tokenizer.pre_tokenizer = Whitespace()
trainer = BpeTrainer(vocab_size=200, special_tokens=["[PAD]","[UNK]","[CLS]","[SEP]","[MASK]"])

tokenizer.train(files=['corpus.txt'], trainer=trainer)

print('Vocab size:', tokenizer.get_vocab_size())
encoded = tokenizer.encode("Generative AI is amazing!")
print('Tokens ->', encoded.tokens)
print('IDs    ->', encoded.ids)
print('Decoded ->', tokenizer.decode(encoded.ids))
```
