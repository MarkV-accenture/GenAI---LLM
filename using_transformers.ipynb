# Using 🤗 Transformers
We'll use the `pipeline` API for common NLP tasks:
- Sentiment analysis
- Text generation
- Translation
We'll also inspect model config and push predictions into a pandas DataFrame.

```python
from transformers import pipeline
import pandas as pd

sentiment_pipe = pipeline('sentiment-analysis')
textgen_pipe = pipeline('text-generation', model='gpt2', max_new_tokens=30)
translate_pipe = pipeline('translation_en_to_fr', model='t5-small')

sample_text = "I love open source AI!"
res_sent = sentiment_pipe(sample_text)
res_gen = textgen_pipe("Once upon a time in Vilnius,")
res_trans = translate_pipe("Transformers make NLP easier.")

df = pd.DataFrame({
    'task':['sentiment','generation','translation'],
    'output':[res_sent, res_gen, res_trans]
})
df
```
