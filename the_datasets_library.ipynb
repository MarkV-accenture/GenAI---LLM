We'll explore, filter, map, and save a dataset.
We'll use the AG News dataset (4-class news topic classification).

```python
from datasets import load_dataset
import pandas as pd

ds = load_dataset('ag_news')
print(ds)

# peek at first rows
print(ds['train'][0])

# simple processing: lowercase text
lower_ds = ds.map(lambda x: {'text': x['text'].lower()})

# train/validation split
split_ds = lower_ds['train'].train_test_split(test_size=0.1, seed=123)
train_ds = split_ds['train']
val_ds = split_ds['test']

# convert small sample to pandas for EDA
df_sample = pd.DataFrame(train_ds.select(range(5)))
df_sample
```

```python
# save processed subset to disk
train_ds.select(range(100)).to_json('ag_news_small_train.json')
val_ds.select(range(100)).to_json('ag_news_small_val.json')
print('Saved ag_news_small_train.json and ag_news_small_val.json')
```
