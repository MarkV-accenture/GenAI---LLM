# Classical NLP Tasks
We'll demo three staple NLP tasks using pretrained models:
1. Text classification (sentiment)
2. Named Entity Recognition (token classification)
3. Question Answering
We'll rely on `pipeline` for simplicity.

```python
!pip install -q transformers torch==2.8.0 --upgrade
```

```python
from transformers import pipeline

# 1. Sentiment classification
sentiment = pipeline('sentiment-analysis')
print('Sentiment:', sentiment("I absolutely loved this product, it works great."))

# 2. Named Entity Recognition (NER)
ner = pipeline('ner', grouped_entities=True)
print('NER:', ner("Hugging Face is based in New York City."))

# 3. Extractive Question Answering
qa = pipeline('question-answering')
context = "Vilnius is the capital and largest city of Lithuania. It is known for its baroque architecture."
question = "What is the capital of Lithuania?"
print('QA:', qa({'question': question, 'context': context}))
```
